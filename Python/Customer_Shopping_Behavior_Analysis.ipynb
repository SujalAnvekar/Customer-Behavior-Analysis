**Handling Null Values**

```python
df['Review Rating']=df.groupby('Category')['Review Rating'].transform(lambda x: x.fillna(x.mean()))
```

```python
df.isnull().sum()
```

**Renaming Columns for furthur Analysis**

```python
df.columns=df.columns.str.lower()
df.columns=df.columns.str.replace(" ","_")
```

```python
df
```

```python
df=df.rename(columns={"purchase_amount_(usd)":"purchase_amt"})
```

```python
df
```

**Featuring Engineering**

```python
labels=['Young Adult','Adult','Middle-aged','Senior']
df['age_group']=pd.qcut(df['age'],q=4,labels=labels)
```

```python
df.columns
```

```python
frequency_mapping={
    'Fortnightly':14,
    'Weekly':7,
    'Monthly':30,
    'Quarterly':90,
    'Annually':365,
    'Bi-Weekly':14,
    'Every 3 Months':90
}
df['purchase_freq_days']=df['frequency_of_purchases'].map(frequency_mapping)
```
